```python
import os
import numpy as np
import pandas as pd
from fredapi import Fred
from dotenv import load_dotenv
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller

load_dotenv()
fred = Fred(api_key=os.getenv('FRED_API_KEY'))

# --- functions (as before) ---

def get_series(ticker, name, filepath):
    if os.path.exists(filepath):
        data = pd.read_csv(filepath, index_col=0, parse_dates=True).squeeze()
        data.name = name
    else:
        data = fred.get_series(ticker)
        data.index = pd.to_datetime(data.index)
        data.name = name
        data.to_csv(filepath, header=True)

    n_missing = data.isna().sum()
    print(f"{name}: {n_missing} missing values")
    if n_missing > 0:
        print(data[data.isna()])

    data = data.ffill().dropna()
    return data


def resample_to_monthly(series, method='mean'):
    if method == 'mean':
        return series.resample('ME').mean()
    elif method == 'last':
        return series.resample('ME').last()
    elif method == 'max':
        return series.resample('ME').max()
    elif method == 'min':
        return series.resample('ME').min()
    else:
        raise ValueError("method must be 'mean', 'last', 'max', or 'min'")


def adf_report(series, name):
    result = adfuller(series.dropna(), autolag='AIC')
    stationary = result[1] < 0.05
    print(f"{name:30s} ADF={result[0]:8.4f}  p-value={result[1]:.4f}  stationary={'YES' if stationary else 'NO'}")
    return stationary, result[0], result[1]


# --- full series config ---
# method: how to collapse to monthly ('mean' for flow-like vars, 'last' for end-of-month snapshots)
# compute_volatility: True only for series where intra-month moves carry separate signal (WTI, BAA10Y)

series_config = {
    'CCSA':      {'name': 'continued_claims', 'method': 'mean', 'compute_volatility': False},
    'ICSA':      {'name': 'initial_claims',   'method': 'mean', 'compute_volatility': False},
    'BAA10Y':    {'name': 'baa10y_spread',    'method': 'last', 'compute_volatility': True},
    'NFCI':      {'name': 'nfci',             'method': 'last', 'compute_volatility': False},
    'T10Y2Y':    {'name': 't10y2y',           'method': 'last', 'compute_volatility': False},
    'CPIAUCSL':  {'name': 'cpi',              'method': 'mean', 'compute_volatility': False},
    'INDPRO':    {'name': 'indpro',           'method': 'mean', 'compute_volatility': False},
    'TB3MS':     {'name': 'tb3ms',            'method': 'mean', 'compute_volatility': False},
    'DCOILWTICO':{'name': 'wti',              'method': 'mean', 'compute_volatility': True},
    'TCU':       {'name': 'tcu',              'method': 'mean', 'compute_volatility': False},
    'UNRATE':    {'name': 'unrate',           'method': 'mean', 'compute_volatility': False},
    'UMCSENT':   {'name': 'umcsent',          'method': 'mean', 'compute_volatility': False},
    'BUSLOANS':  {'name': 'busloans',         'method': 'mean', 'compute_volatility': False},
    'CUMFNS':    {'name': 'cumfns',           'method': 'mean', 'compute_volatility': False},
    'M2SL':      {'name': 'm2',               'method': 'mean', 'compute_volatility': False},
    'PAYEMS':    {'name': 'payems',           'method': 'mean', 'compute_volatility': False},
    'PPIACO':    {'name': 'ppiaco',           'method': 'mean', 'compute_volatility': False},
}

raw_series = {}
monthly_series = {}
volatility_series = {}
adf_results = {}

for ticker, cfg in series_config.items():
    name = cfg['name']
    filepath = f"../data/{name}.csv"

    # --- pull raw ---
    raw = get_series(ticker, name, filepath)
    raw_series[ticker] = raw

    # --- resample to monthly ---
    monthly = resample_to_monthly(raw, method=cfg['method'])
    monthly.name = f"{name}_monthly"
    monthly_series[ticker] = monthly
    monthly.to_csv(f"../data/processed/{name}_monthly.csv", header=True)

    # --- plot ---
    print(f"--- {ticker}: date range {raw.index.min().date()} to {raw.index.max().date()}, "
          f"{len(raw)} raw obs, {len(monthly)} monthly obs ---")

    plt.figure(figsize=(10, 4))
    plt.plot(monthly)
    plt.title(f'{name} (Monthly)')
    plt.show()

    # --- ADF test on monthly level ---
    is_stationary, adf_stat, p_value = adf_report(monthly, monthly.name)
    adf_results[ticker] = {
        'level_stationary': is_stationary,
        'level_adf_stat': adf_stat,
        'level_p_value': p_value,
    }

    # if not stationary at level, check first difference too — flag, don't auto-replace
    if not is_stationary:
        diff = monthly.diff().dropna()
        diff_stationary, diff_stat, diff_p = adf_report(diff, f"{name}_monthly_diff")
        adf_results[ticker]['diff_stationary'] = diff_stationary
        adf_results[ticker]['diff_adf_stat'] = diff_stat
        adf_results[ticker]['diff_p_value'] = diff_p

    # --- optional monthly volatility (only for flagged series) ---
    if cfg['compute_volatility']:
        daily_logret = np.log(raw).diff()
        vol_monthly = daily_logret.resample('ME').std()
        vol_monthly.name = f"{name}_volatility_monthly"
        volatility_series[ticker] = vol_monthly
        vol_monthly.to_csv(f"../data/processed/{name}_volatility_monthly.csv", header=True)

        plt.figure(figsize=(10, 4))
        plt.plot(vol_monthly)
        plt.title(f'{name} — Monthly Realized Volatility (std. of daily log returns)')
        plt.show()

        vol_stationary, vol_stat, vol_p = adf_report(vol_monthly.dropna(), vol_monthly.name)
        adf_results[ticker]['volatility_stationary'] = vol_stationary


# --- summary of all stationarity flags ---
print("\n--- ADF summary: series requiring attention (not stationary at level) ---")
for ticker, res in adf_results.items():
    if not res['level_stationary']:
        print(f"{ticker}: level NOT stationary (p={res['level_p_value']:.4f}); "
              f"first diff stationary = {res.get('diff_stationary', 'not tested')}")
```
