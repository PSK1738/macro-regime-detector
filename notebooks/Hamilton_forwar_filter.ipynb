# Hamilton (1989) Forward Filter

In order to find expansion/recession probabilities over time, we will use Hamilton's 2 state model, his MLE's for the parameters of the normally distributed GDP growth under each state.


| Parameter | Value |
|-----------|-------|
| $\mu_0$ (expansion) | 1.164 |
| $\mu_1$ (recession) | -0.358 |
| $\sigma$ | 1.123 |
| $p_{00}$ | 0.9049 |
| $p_{11}$ | 0.7550 |

Initial state vector: $ x_0 = [0.8, 0.2]$

## The Filter

**Step 1 — Prediction**

$$ /hat(x_t) = x_{t-1} P$$

**Step 2 — Update**

$$x_t = \frac{w_t \odot \dot{x}_t}{\mathbf{1}^\top (w_t \odot \dot{x}_t)}$$

Where $w_t = [f(y_t \mid S_t = 0),\ f(y_t \mid S_t = 1)]$ is the vector of state-conditional densities and $\odot$ is element-wise multiplication. The denominator is just the sum of the numerator, normalising so probabilities sum to 1.
